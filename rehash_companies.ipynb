```python
aug_companies = predated_companies[(predated_companies['updated_at'] >= '2024-08-01') & (predated_companies['updated_at'] <= '2024-08-31')].reset_index(drop=True).copy()
```

```python
aug_companies
```

```python
headcount_cols = [f'headcount_{i}m' for i in range(0,19)]
```

```python
def shift_headcount(row, shift):
    max_shift = 18 - shift
    trailing_shift = 12 - shift
    for i in range(0, max_shift + 1): 
        row[f"headcount_{i+shift}m"] = row[f"headcount_{i}m"]
        row[f"linkedin_follower_count_{i+shift}m"] = row[f"linkedin_follower_count_{i}m"]
        row[f"twitter_follower_count_{i+shift}m"] = row[f"twitter_follower_count_{i}m"]

    for i in range(0, trailing_shift + 1): 
        row[f"headcount_growth_trailing_3m_{i+shift}m"] = row[f"headcount_growth_trailing_3m_{i}m"]

    for i in range(0, shift):
        row[f"headcount_{i}m"] = np.nan
        row[f"headcount_growth_trailing_3m_{i}m"] = np.nan
        row[f"linkedin_follower_count_{i}m"] = np.nan
        row[f"twitter_follower_count_{i}m"] = np.nan
    return row

aug_companies = aug_companies.apply(shift_headcount, args=(4,), axis=1)
```

```python
oct_companies = predated_companies[(predated_companies['updated_at'] >= '2024-10-01') & (predated_companies['updated_at'] <= '2024-10-31')].reset_index(drop=True).copy()
```

```python
oct_companies[headcount_cols]
```

```python
oct_companies = oct_companies.apply(shift_headcount, args=(2,), axis=1)
```

```python
oct_companies[headcount_cols]
```

```python
nov_companies = predated_companies[(predated_companies['updated_at'] >= '2024-11-01') & (predated_companies['updated_at'] <= '2024-11-30')].reset_index(drop=True).copy()
```

```python
nov_companies = nov_companies.apply(shift_headcount, args=(1,), axis=1)
```

```python
nov_companies[headcount_cols]
```

```python
predated_companies_rehashed = pd.concat([aug_companies, oct_companies, nov_companies])
```

```python
predated_companies_rehashed = predated_companies_rehashed.drop(columns=['headcount_18m', 'headcount_growth_trailing_3m_12m',
                                          "linkedin_follower_count_18m", "twitter_follower_count_18m"])
```

```python
predated_companies_rehashed.to_csv('predated_companies_rehashed.csv', index=False)
```
